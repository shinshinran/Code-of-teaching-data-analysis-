```python
#生成一个分界
bins = [0, 5, 10, 25, 50, 100, 1951]
#分组并画图
pd.cut(user_df['浏览量'], bins, right=True,labels=[f'({bins[x]}，{bins[x+1]}]' for x in range(len(bins)-1)]).value_counts().plot(kind='pie')
```

```python
#或者画一个环形图
#先把数据放进data
data=pd.cut(user_df['浏览量'], bins, right=True,labels=[f'({bins[x]}，{bins[x+1]}]' for x in range(len(bins)-1)]).value_counts()
#然后用wedgeprops=dict(width=0.3, edgecolor='w')参数控制画环形图

plt.pie(data,labels=data.index,wedgeprops=dict(width=0.3, edgecolor='w'))

plt.show()
```
