```python
import pandas as pd
import re
import numpy as np
```

```python
df=pd.read_csv(r'cdlj.csv')
```

```python
pattern = re.compile( r'\b(东南|东北|西南|西北|东|南|西|北)\b')
match = pattern.match(df['HouseIcon'][0])
```

```python
pattern = r'(?:.*\|)?(.*)'

# 使用re.search查找匹配项
match = re.search(pattern, df['HouseIcon'][0])
```

```python
match.group(1)
```

```python
df['room']=df['HouseIcon'].apply(lambda x :re.search(r'\d+室\d+厅',x).group(0))

df['area']=df['HouseIcon'].apply(lambda x :re.search(r'(\d+\.?\d*)\s*平米',x).group(0))

df['direct']=df['HouseIcon'].apply(lambda x :re.search( r'\b(东南|东北|西南|西北|东|南|西|北)\b',x).group(0))

df['type']=df['HouseIcon'].apply(lambda x :re.search( r'(?:\|.*?){2}\|(.+?)\|',x).group(1))

df['floor']=df['HouseIcon'].apply(lambda x :re.search( r'(?:\|.*?){3}\|(.+?)\|',x).group(1))
```

```python
def yr(a):
    pattern =r'\b\d{4}年\b'
    match = re.search(pattern, df['HouseIcon'][0])
    if match is None:
        return np.nan
    else:
        return match.group(0)
```

```python
df['year']=df['HouseIcon'].apply(lambda x :yr(x))
```

```python
df['struct']=df['HouseIcon'].apply(lambda x :re.search(r'(?:.*\|)?(.*)',x).group(1))
```
