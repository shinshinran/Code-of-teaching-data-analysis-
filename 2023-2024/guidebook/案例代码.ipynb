```python
def xprnc(x):
    if re.search('\d-\d年',x) is not None:
           return re.search('\d-\d年',x)[0]
    elif re.search('\d个月',x) is not None:
           return re.search('\d个月',x)[0]
    elif re.search('(?:\d{1,2}个月|\d{1,2}年以内|\d{1,2}年以上)',x) is not None :
           return re.search('(?:\d{1,2}个月|\d{1,2}年以内|\d{1,2}年以上)',x)[0]
    elif re.search('经验不限',x)is not None:
           return 0
           
        
```

```python
df['经验']=df['学位'].apply(lambda x:xprnc(x))
df
```

```python
def schlor(x):
    if re.search('中专/中技 ',x) is not None:
        return re.search('中专/中技 ',x)[0]
    elif re.search('大专',x) is not None:
        return re.search('大专',x)[0]
    elif re.search('高中',x) is not None:
        return re.search('高中',x)[0]
    elif re.search('本科',x) is not None:
        return re.search('本科',x)[0]
    elif re.search('硕士',x) is not None:
        return re.search('硕士',x)[0]
    elif re.search('博士',x) is not None:
        return re.search('博士',x)[0]
    elif re.search('学历不限',x) is not None:
        return re.search('学历不限',x)[0]
    
```

```python
df['学位1']=df['学位'].apply(lambda x:schlor(x))
df
```

```python
def lower(x):
    match = re.search(r'(\d+)-(\d+)K', x)
    if re.search('(?<=·)\d+',x) is not None:
        if match:
            return (int(match.group(1)) * 1000)*(int(re.search('(?<=·)\d+',x)[0]))
    else:
        if match:
            return (int(match.group(1)) * 1000)*12

```

```python
df2['lower']=df2['薪资'].apply(lambda x: lower(x))
df2.head()
```

```python
def upper(x):
    match = re.search(r'(\d+)-(\d+)K', x)
    if re.search('(?<=·)\d+',x) is not None:
        if match:
            return (int(match.group(2)) * 1000)*(int(re.search('(?<=·)\d+',x)[0]))
    else:
        if match:
            return (int(match.group(2)) * 1000)*12
```

```python
df2['upper']=df2['薪资'].apply(lambda x: upper(x))
df2.head()
```

```python
#merged_string = df['职位要求'].agg(lambda x: ','.join(x.split(',')))
# 假设df是你的DataFrame，'column_name'是你想要合并的那一列
merged_string = df['职位要求'].apply(lambda x: ','.join(x.split(','))).str.cat(sep=',')
```

```python
merged_string.split().explode().value_counts()
```

```python
merged_string
```

```python

# 使用逗号分割字符串，得到单词列表
words = merged_string.split(',').split('，')

# 创建一个空字典用于存储单词频数
word_count = {}

# 统计每个单词的频数
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# 打印每个单词的频数
for word, count in word_count.items():
    print(f"{word}: {count}")
```

```python
import re

def calculate_salary_range(salary_str):
    # 使用正则表达式匹配数字
    match = re.match(r'(\d+)-(\d+)K', salary_str)
    if match:
        lower_bound = int(match.group(1)) * 1000  # 转换为实际薪资并乘以1000
        upper_bound = int(match.group(2)) * 1000  # 同上
        return lower_bound, upper_bound
    else:
        raise ValueError("Invalid salary format")

# 测试函数
salary_str = "20-40K"
lower, upper = calculate_salary_range(salary_str)
print(f"Salary range: {lower} - {upper}")
```
